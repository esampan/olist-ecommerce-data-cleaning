## 2. Data Cleaning — Orders Dataset

In this section, I will profile, clean, validate, and prepare the Olist orders dataset for analysis.

```python
## 2. Data Cleaning — Orders Dataset

## Load the orders table Dataset

orders = pd.read_csv(
    '/content/drive/MyDrive/olist_dataset/2 olist_orders_dataset.csv'
)

orders.head()
```

```python
## Check orders table Dataset Shape

orders.shape
```

```python
## Check Orders Dataset types

orders.dtypes
```

```python
## Checking for Missing Values

orders.isnull().sum()
```

```python
## Checking if the missing dates correspond to particular order statuses.

orders['order_status'].value_counts()
```

```python
## Check missing date values for each order status

orders.groupby('order_status')[ ['order_approved_at', 'order_delivered_carrier_date',
        'order_delivered_customer_date'] ].apply(lambda x: x.isnull().sum())
```

```python
## Find delivered orders with missing delivery-related dates

orders[(orders['order_status'] == 'delivered') &(orders[['order_approved_at','order_delivered_carrier_date',
                'order_delivered_customer_date']].isnull().any(axis=1))]
```

```python
## Checking for sum of unique order_id

orders['order_id'].duplicated().sum()
```

```python
## Check for duplicate rows in the orders dataset

orders.duplicated().sum()
```

```python
## Check the unique order status values

orders['order_status'].unique()
```

```python
## Check each date column for invalid or unrecognized date values

date_columns = [
    'order_purchase_timestamp',
    'order_approved_at',
    'order_delivered_carrier_date',
    'order_delivered_customer_date',
    'order_estimated_delivery_date']

for col in date_columns:
    invalid_dates = pd.to_datetime(orders[col], errors='coerce').isna().sum()
    print(f"{col}: {invalid_dates} invalid dates")
```

```python
## Convert all date columns to datetime format

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
```

```python
## Check data types and confirm that date columns are in datetime format

orders.dtypes
```

```python
# Save the cleaned orders dataset to Google Drive
cleaned_path = '/content/drive/MyDrive/olist_dataset/cleaned_olist_orders_dataset.csv'

orders.to_csv(cleaned_path, index=False)

print("Saved successfully!")
print(cleaned_path)
```
